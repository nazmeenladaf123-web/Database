1. Explain the fundamental differences between DDL, DML, and DQL commands in SQL. Provide one example for each type of command.

Ans :- In SQL, DDL, DML, and DQL are three fundamental categories of commands that serve different purposes in managing and interacting with a database. DDL (Data Definition Language) is used to define and modify the structure of database objects such as tables, schemas, and indexes. It deals with the design of the database itself rather than the data inside it; for example, a command like CREATE TABLE Students (ID INT, Name VARCHAR(50)); is a DDL statement because it creates a new table structure. DML (Data Manipulation Language) is used to insert, update, or delete the actual data stored in the database tables. It focuses on modifying the records within existing structures, such as INSERT INTO Students VALUES (1, 'Amit');, which adds a new row of data to the table. DQL (Data Query Language) is used to retrieve data from the database without changing it, and its primary command is SELECT, for instance SELECT * FROM Students;, which fetches all records from the Students table.

2.  What is the purpose of SQL constraints? Name and describe three common types
of constraints, providing a simple scenario where each would be useful.

Ans :- The purpose of SQL constraints is to enforce rules on the data stored in database tables to ensure accuracy, consistency, and reliability. They help maintain data integrity by restricting the type of data that can be inserted, updated, or deleted in a table.

Three common constraints are NOT NULL, UNIQUE, and PRIMARY KEY. The NOT NULL constraint ensures that a column cannot have empty values; for example, in a Students table, the “Name” column can be set as NOT NULL so every student must have a name entered. The UNIQUE constraint ensures that all values in a column are different from each other; for instance, an “Email” column in a Users table can use UNIQUE so no two users can register with the same email address. The PRIMARY KEY constraint uniquely identifies each record in a table and does not allow duplicate or NULL values; for example, a “StudentID” column in a Students table can be defined as a primary key so each student can be uniquely identified.

3. Explain the difference between LIMIT and OFFSET clauses in SQL. How
would you use them together to retrieve the third page of results, assuming each page
has 10 records?

Ans :- The LIMIT and OFFSET clauses in SQL are used for pagination, which means retrieving query results in smaller, manageable chunks instead of fetching all records at once. The LIMIT clause specifies the maximum number of rows to return, while the OFFSET clause specifies how many rows to skip before starting to return results.

When used together, they allow controlled navigation through pages of data. For example, to retrieve the third page of results where each page contains 10 records, you first need to skip the first 20 records (because page 1 has records 1–10 and page 2 has records 11–20), and then fetch the next 10 records.

 The SQL query would be:
SELECT *
FROM table_name
LIMIT 10 OFFSET 20;

Here, OFFSET 20 skips the first two pages (20 records), and LIMIT 10 returns the next 10 records, which represent the third page of results.

4. What is a Common Table Expression (CTE) in SQL, and what are its main
benefits? Provide a simple SQL example demonstrating its usage.

Ans :- A Common Table Expression (CTE) in SQL is a temporary, named result set that you can define within a query using the WITH clause. It exists only during the execution of that query and can be referenced like a table. CTEs are mainly used to simplify complex queries by breaking them into readable, reusable parts, especially when dealing with joins, subqueries, or recursive logic.

The main benefits of a CTE are improved readability, easier maintenance, and better query structure. Instead of writing deeply nested subqueries, you can separate logic into clear steps. CTEs also make debugging easier because each part of the query is logically organized. Additionally, recursive CTEs allow querying hierarchical data like employee-manager relationships or tree structures.

For example, suppose we want to find employees with a salary greater than the average salary:

WITH AvgSalary AS (
    SELECT AVG(salary) AS avg_sal
    FROM Employees
)
SELECT name, salary
FROM Employees, AvgSalary
WHERE Employees.salary > AvgSalary.avg_sal;

Here, the CTE AvgSalary calculates the average salary first, and the main query then uses it to filter employees earning above that average.

5.  Describe the concept of SQL Normalization and its primary goals. Briefly
explain the first three normal forms (1NF, 2NF, 3NF).

ANS :- SQL Normalization is the process of organizing data in a relational database to reduce redundancy and improve data integrity. It involves dividing large tables into smaller, related tables and defining relationships between them. The primary goals of normalization are to eliminate duplicate data, ensure consistent data storage, and make the database easier to maintain and update without anomalies like insertion, update, and deletion issues.

The First Normal Form (1NF) ensures that each column contains only atomic (indivisible) values and that each record is unique. In simple terms, a table should not have repeating groups or multiple values in a single column; for example, instead of storing multiple phone numbers in one field, each phone number should be stored in a separate row or column.

The Second Normal Form (2NF) builds on 1NF and requires that all non-key attributes are fully dependent on the entire primary key, not just part of it. This mainly applies to tables with composite primary keys; for example, in an order details table with (OrderID, ProductID) as a key, attributes like ProductName should depend only on ProductID and be moved to a separate Product table.

The Third Normal Form (3NF) further refines 2NF by ensuring that there are no transitive dependencies, meaning non-key attributes should not depend on other non-key attributes. For example, if a table contains EmployeeID, DepartmentID, and DepartmentName, then DepartmentName should be moved to a separate Department table because it depends on DepartmentID rather than directly on EmployeeID.
